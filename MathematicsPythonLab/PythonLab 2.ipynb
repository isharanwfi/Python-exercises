```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,4)
y = 0.1*x**4 - 0.5*x**3 + 0.2*x**2 - 0.1*x + 4

fig, ax = plt.subplots(figsize=(10,6))
plt.plot(x,y, linestyle='--', linewidth = 2.5, label="4th order polynomial function")
plt.grid(True)
plt.xlabel("X", fontsize=16)
plt.ylabel("Y", fontsize=16)
plt.title(" Graph of a function", fontsize=19)
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots()

x = np.linspace(0,4)
y1 = x-2
y2 = x**2 - 4*x

plt.plot(x,y1, label="$ y = x-2 $")
plt.plot(x,y2, label="$ y = x^2 - 4x $")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(2, figsize = (10,8))

x = np.linspace(0,5)
y1 = x-2
plt.sca(ax[0])
plt.plot(x,y1, colour= "b", label= "x-2")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()


```
